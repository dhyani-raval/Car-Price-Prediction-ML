__Business Problem :-__ _We have to create a model which can predict the selling price of Car._

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('car data.csv')
```

```python
df.head()
```

__Data Preprocessing__

```python
df.isnull().sum()
```

```python
df.drop("Car_Name",axis=1,inplace=True)
```

```python
df.head(3)
```

```python
df["Year"] = 2026-df.Year
```

```python
df.head(3)
```

```python
df = pd.get_dummies(df,dtype=int,drop_first=True)
```

```python
df.head()
```

```python
X = df.drop("Selling_Price",axis=1)
y = df.Selling_Price
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=9)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf = RandomForestRegressor()
rf.fit(X_train,y_train) # train
y_pred = rf.predict(X_test)  # prediction
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
print(mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))   # testing performance
```

```python
y_pred_train = rf.predict(X_train)
r2_score(y_train,y_pred_train)   # training performance
```
